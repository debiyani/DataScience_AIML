### Lab Question 3: 
    Student Database Update
    Create a dictionary named student with the following keys: "name": "Alice", "age": 20, and "major": "Physics". 
    1. Add a new key-value pair: "gpa": 3.8. 
    2. Update the "major" to "Computer Science". 
    3. Delete the "age" key using the pop() method. 
    4. Loop through the dictionary and print only the keys. 
    5. Loop through the dictionary using .items() and print the details in the format: Field: 
    Value.

```python
student={
    'name':'Alice',
    'age':20,
    'major':'Physics'
    }
student['major']='Computer Science'
student.pop('age')
for key in student.keys():
    print(key)
for key, val in student.items():
    print(f"{key}: {val}")
```

### Lab Question 4: 
    Nested Data & The "Menu" Challenge 
    1. Create a dictionary called cafeteria_menu where the keys are categories (e.g., "Snacks" "Drinks") and the values are lists of items. 
    2. Add a new item to the "Drinks" list using the .append() method inside the dictionary. 
    3. Use a for loop and the enumerate() function to print the list of "Snacks" with their index numbers (starting from 1). 
    4. Access and print the second item from the "Snacks" list using double indexing (e.g., dict[key][index]).

```python
cafeteria_menu={
    "Snacks":["Cookie","Chips", "Chocolate"],
    "Drinks":["Coke","Juice"]
}
cafeteria_menu["Drinks"].append("Coffee")
for index, value in enumerate(cafeteria_menu["Snacks"]):
    if index>0:
        print(f"{index}: {value}")
print(cafeteria_menu['Snacks'][1])
```
